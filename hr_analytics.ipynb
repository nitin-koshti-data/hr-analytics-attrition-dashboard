```python
import numpy as np
import pandas as pd

df=pd.read_csv('HR_data.csv')
df.head(5)
```

```python
print("shape :",df.shape)
print("--------------------------------------")
print(df.info())
print("--------------------------------------")
print("Nulls :\n",df.isnull().sum())
print("--------------------------------------")
print("Duplicate :",df.duplicated().sum())
```

```python
df.columns=df.columns.str.strip().str.replace(" ","_")
```

```python
#drop useless column
#employee count column is 1 for every row -so no need to analytics use so drop it
if 'Employee_Count' in df.columns:
    df=df.drop(columns=['Employee_Count'])
print(df.columns)
```

```python
# create new column income level
df['Income_Level']=pd.qcut(df['Monthly_Income'],q=4,labels=['Low','Medium','High','Very High'])
```

```python
# tenure buckets
df['Tenure_Group']=pd.cut(df['Years_At_Company'],bins=[-1,2,5,10,20,100],labels=['0-2 years','3-5 years','6-10 years','11-20 years','20+ years'])
```

```python
# Map numeric satisfaction to label
satisfaction_map={1:'Low',2:'Medium',3:'High',4:'Very High'}
df['Job_Satisfaction']=df['Job_Satisfaction'].map(satisfaction_map)
df['Environment_Satisfaction']=df['Environment_Satisfaction'].map(satisfaction_map)
```

```python
#promotion risk flag -- year since last promotion vs years at a company
df['Promotion_Overdue']=(df['Years_Since_Last_Promotion']>=5).astype(int)
df[['Years_Since_Last_Promotion','Promotion_Overdue']]
```

```python
df['attrition_flag']=df['attrition'].map({'Yes':1,'No':0})
```

```python
print("\n Cleaned shape:",df.shape)
print("--------------------------------------")
print(df.head())
print("--------------------------------------")
print(df.dtypes)
```

```python
df.to_csv("HR_Attrition_cleaned.csv",index=False)
```
